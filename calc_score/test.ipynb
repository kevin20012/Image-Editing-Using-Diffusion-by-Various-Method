```python
from PIL import Image

# 1. 이미지 불러오기 (L 모드로 변환)
image_path = "./img/000000000000_mask.png"
image = Image.open(image_path).convert("L")

# 2. 알파 채널 생성: 픽셀 값이 0이면 0(투명), 1이면 255(불투명)으로 매핑
alpha = image.point(lambda p: 0 if p == 0 else 255)

# 3. 원본 그레이스케일 이미지를 RGB로 변환 (여기서는 기존 밝기를 유지)
rgb = image.convert("RGB")

# 4. RGB 이미지에 알파 채널을 추가하여 RGBA 이미지 생성
im_rgba = rgb.copy()
im_rgba.putalpha(alpha)

im_rgba
```
